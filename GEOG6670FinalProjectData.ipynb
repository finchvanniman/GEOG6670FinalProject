# Data File Links

The Earth Engine datasets were pulled in through gee. The Social Vulnerability Index (SVI) data was downloaded as a CSV and joined to the counties. The remaining environmental data was downloaded as a geojson and spatially joined to the counties and SVI data. 

Google Earth Engine Population Density: https://developers.google.com/earth-engine/datasets/catalog/CIESIN_GPWv411_GPW_Population_Density

Google Earth Engine US Census Counties 2018: https://developers.google.com/earth-engine/datasets/catalog/TIGER_2018_Counties#table-schema

I had initially wanted to convert a geojson or a shape file into an Earth Engine format, however, I continued to recieve an error that the file was too large and exceeded the limits. Due to this, I opted to pull a second Earth Engine Feature Collection to utilize as my counties.

Social Vulnerability Index Data by County for 2022: https://www.atsdr.cdc.gov/place-health/php/svi/svi-data-documentation-download.html

Hazardous Waste Treatment, Storage, and Disposal Facilities: https://gis-michigan.opendata.arcgis.com/datasets/eddb0a8743e640928d6a1a9fe6277b94_7/explore?location=44.907425%2C-86.103924%2C6

Solid Waste Landfills: https://gis-michigan.opendata.arcgis.com/datasets/43afb115983b4c62900c7ab129e0a3e0_6/explore?location=44.907425%2C-86.103924%2C6

Environmental Contamination Sites: https://gis-michigan.opendata.arcgis.com/datasets/2bb7f2d4a9b4459bac03d7f73a7fc399_0/explore?location=44.809768%2C-86.388492%2C6
