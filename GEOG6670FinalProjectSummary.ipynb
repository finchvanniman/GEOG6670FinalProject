# GEOG6670 Final Project Summary and Notes - Vanniman

This project aimed to create an environmental risk score for Michigan counties based on hazardous waste treatment and storage locations, solid waste landfills, environmental contamination sites, and PFAS sites of interest. Then, the overall Social Vulnerability (SVI) Theme scores were added to the risk score as these can compound the effects felt in the area. Social vulnerability is are the demographic and socioeconomic variables that influce how an inividual or community responds to a disaster or other major event. Further, population averages were added to the risk score to include the number of people living in the area that the risk score impacts. The more people living in an area, the more of an imapct the risk score has on humans.
